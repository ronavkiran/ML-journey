```python
import torch
from sklearn.datasets import fetch_california_housing
X,y=fetch_california_housing(return_X_y=True)
X_train=torch.tensor(X,dtype=torch.float32,device="cuda")
y_target=torch.tensor(y,dtype=torch.float32,device="cuda").unsqueeze(1)
xmean=X_train.mean(dim=0,keepdim=True)
xstd=X_train.std(dim=0,keepdim=True)
print(X_train.shape)
print(y_target.shape)

#normalizing the data
X_train=(X_train-xmean)/xstd

torch.manual_seed(24)
features=X_train.shape[1]
w=torch.randn((features,1),requires_grad=True,device="cuda")
b=torch.zeros(1,requires_grad=True,device="cuda")
print(w.shape)
print(b.shape)

print(w)
```

```python
learning_rate=0.1
for i in range(100):
    #foward pass
    y_pred=X_train@w+b
    loss=((y_target-y_pred)**2).mean()
    #backward pass
    loss.backward()

    #gradient descent
    with torch.no_grad():
        w-=learning_rate*w.grad
        b-=learning_rate*b.grad
    if i%10==0:
        print(loss.item()) 
    w.grad.zero_()
    b.grad.zero_()
```
