```python
import torch
from sklearn.datasets import load_diabetes
X,y=load_diabetes(return_X_y=True)
X_train=torch.tensor(X,dtype=torch.float32,device="cuda")
y_target=torch.tensor(y,dtype=torch.float32,device="cuda").unsqueeze(1)
print(X_train.shape)
print(y_target.shape)

#Normalizing
xmean=X_train.mean(dim=0)
xstd=X_train.std(dim=0)
X_train=(X_train-xmean)/xstd

#Normalize the target to get better MSE
ymean=y_target.mean()
ystd=y_target.std()
y_target=(y_target-ymean)/ystd

features=X_train.shape[1]
torch.manual_seed(67)
w=torch.randn((features,1),requires_grad=True,device="cuda",dtype=torch.float32)
b=torch.zeros(1,requires_grad=True,device="cuda",dtype=torch.float32)
```

```python
learning_rate=0.1
for i in range(100):
    #foward
    y_pred=X_train@w+b
    loss=((y_target-y_pred)**2).mean()

    #backward
    loss.backward()

    #gradient descent
    with torch.no_grad():
        w-=learning_rate*w.grad
        b-=learning_rate*b.grad
    w.grad.zero_()
    b.grad.zero_()

    if i%10==0:
        print(loss.item())
```
